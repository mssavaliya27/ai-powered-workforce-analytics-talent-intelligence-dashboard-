```python
import sqlite3

# Connect to database
conn = sqlite3.connect("student.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Students (
    ID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER
)
""")

# Insert data
cursor.execute("INSERT INTO Students VALUES (1, 'Rahul', 20)")
cursor.execute("INSERT INTO Students VALUES (2, 'Priya', 21)")
cursor.execute("INSERT INTO Students VALUES (3, 'Amit', 22)")

# Save changes
conn.commit()

# Display data
print("Student Records")

cursor.execute("SELECT * FROM Students")
rows = cursor.fetchall()

for row in rows:
    print(row)

# Close connection
conn.close()
```

```python
import sqlite3

# Database બનાવો
conn = sqlite3.connect("student.db")
cur = conn.cursor()

# Table બનાવો
cur.execute("""
CREATE TABLE IF NOT EXISTS Student (
    ID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER
)
""")

# Data ઉમેરો
cur.execute("INSERT INTO Student VALUES (1, 'Rahul', 20)")
cur.execute("INSERT INTO Student VALUES (2, 'Priya', 21)")
cur.execute("INSERT INTO Student VALUES (3, 'Amit', 22)")

conn.commit()

# Data બતાવો
print("Student Records:")
for row in cur.execute("SELECT * FROM Student"):
    print(row)

conn.close()
```
